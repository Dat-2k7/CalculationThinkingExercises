#### Bài 2B: Viết hàm zigzag_numbers(n) trả về một chuỗi biểu diễn ma trận n x n chứa các số từ 1 đến n^2. Các hàng cách nhau bởi ký tự xuống dòng. Với chỉ số hàng bắt đầu từ 0: hàng chẵn đi từ trái sang phải, hàng lẻ đi từ phải sang trái. Các số trong cùng một hàng cách nhau bởi một dấu cách. 

```python
def zigzag_numbers(n):
    ans = []
    curr = 1
    for i in range(n):
        row = []
        for j in range(n):
            row.append(curr)
            curr += 1
        if i % 2 != 0:
            row.reverse()
        ans.append(row)
    for i in ans:
        print(*i)
n = int(input())
zigzag_numbers(n)
```

#### Bài 3: Trả về cột (chỉ số bắt đầu từ 0) có tổng cột lớn nhất trong ma trận n x m. Viết hàm column_with_max_sum(mat) trả về chỉ số bắt đầu từ 0 của cột có tổng lớn nhất. Nếu có nhiều cột cùng tổng lớn nhất, trả về cột có chỉ số nhỏ nhất

```python
def column_with_max_sum(m):
    s = []
    rows = len(m)
    column = len(m[0])
    for j in range(column):
        column_sum = 0
        for i in range(rows):
            column_sum += m[i][j]
        s.append(column_sum)
    max_v = max(s)
    for i, x in enumerate(s):
        if x == max_v:
            return i
n, m = list(map(int, input().split()))
res = []
for i in range(n):
    res.append(list(map(int, input().split())))
print(column_with_max_sum(res))
```

#### Bài 4: Trả về hàng (chỉ số bắt đầu từ 0) có tổng hàng nhỏ nhất trong ma trận n x m. Viết hàm row_with_min_sum(mat) trả về chỉ số bắt đầu từ 0 của hàng có tổng nhỏ nhất. Nếu có nhiều hàng cùng tổng nhỏ nhất, trả về hàng có chỉ số lớn nhất

```python
def column_with_min_sum(mat):
    s = []
    for j in range(len(mat[0])):
        column_sum = 0
        for i in range(len(mat)):
            column_sum += mat[i][j]
        s.append(column_sum)
    for i, x in enumerate(s):
        if x == min(s):
            return i
n, m = list(map(int, input().split()))
ans = []
for i in range(n):
    ans.append(list(map(int,input().split())))
column_with_min_sum(ans)
```

#### Bài 5: Cho ma trận vuông cấp n. In ra đường chéo chính và đường chéo phụ của ma trận

```python
n = int(input())
ans = []
for i in range(n):
    curr = list(map(int, input().split()))
    ans.append(curr)
cheo_chinh = [ans[i][i] for i in range(n)]
cheo_phu = [ans[i][n - i - 1] for i in range(n)]
print(*(cheo_chinh))
print(*(cheo_phu))
```

#### Bài 6: Cho ma trận nxm số nguyên, thêm vào một cột tại vị trí k.

```python
def insert_column(matrix, k, column):
    rows = len(matrix)
    for i in range(rows):
        matrix[i].insert(k, column[i])
    return matrix
n, m = list(map(int, input().split()))
matrix = []
for i in range(n):
    matrix.append(list(map(int, input().split())))
k = int(input())
column = []
for i in range(n):
    column.append(int(input()))
result_matrix = insert_column(matrix, k, column)
print("Ma trận sau khi thêm cột:")
for row in result_matrix:
    print(*(row))
```

#### Mở rộng: Bài toán tìm điểm yên ngựa: Cho một ma trận hai chiều kích thước $n \times m$ gồm các số nguyên. Một phần tử trong ma trận được gọi là điểm yên ngựa (Saddle Point) nếu nó thỏa mãn đồng thời hai điều kiện sau:Có giá trị nhỏ nhất trên hàng chứa nó.Có giá trị lớn nhất trên cột chứa nó.Hãy viết chương trình tìm tất cả các điểm yên ngựa có trong ma trận đã cho. Nếu ma trận không có điểm yên ngựa nào, in ra thông báo tương ứng

```python
def diem_yen_ngua(matrix):
    n = len(matrix)
    m = len(matrix[0])
    lst = []
    for i in range(n):
        min_row_val = min(matrix[i])
        for j in range(m):
            if matrix[i][j] == min_row_val:
                is_diem_yen_ngua = True
                for k in range(n):
                    if matrix[k][j] > min_row_val:
                        is_diem_yen_ngua = False
                        break
                if is_diem_yen_ngua:
                    lst.append((i, j, min_row_val))
    return lst
n, m = list(map(int, input().split()))
matrix = []
for i in range(n):
    matrix.append(list(map(int, input().split())))
results = diem_yen_ngua(matrix)
if len(results) == 0:
    print("Không có điểm yên ngựa nào.")
else:
    for row in matrix:
        print(*row)
    for point in results:
        print(f"Vị trí ({point[0]}, {point[1]}) có giá trị: {point[2]}")
```
