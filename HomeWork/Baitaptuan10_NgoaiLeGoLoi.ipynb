#### W10A1 DivideByZero: Nhập vào hai số nguyên $a, b$. Thực hiện in ra kết quả phép chia $a/b$. Bắt lỗi ngoại lệ khi $b = 0$. Kết quả in ra chính xác tới 2 chữ số thập phân nếu $b \neq 0$. (Dùng try...except bắt lỗi ZeroDivisionError)

```python
try:
    a = int(input())
    b = int(input())
    ans = a / b
    print(f"{ans:.2f}")
except ZeroDivisionError:
    print("Lỗi chia cho 0!")
```

#### W10A2 Invalid_Index: Hãy nhập vào một mảng my_list (list) $n$ số nguyên ($n$ nguyên dương). Thực hiện thử lệnh print(my_list[n]) rồi đọc kỹ Traceback và bắt lỗi Truy cập chỉ số sai. (Dùng try...except bắt lỗi IndexError).

```python
try:
    n = int(input())
    my_list = []
    for i in range(n):
        my_list.append(int(input()))
    print(my_list[n])
except IndexError:
    print("Lỗi: Truy cập chỉ số vượt quá phạm vi của danh sách!")
```

#### W10A3 ValueError - HappyNumbers: Nhập vào một số nguyên dương $a$, Kiểm tra xem $a$ có phải là số hạnh phúc không.Một số nguyên dương $n$ được gọi là số hạnh phúc nếu ta thực hiện liên tiếp các bước sau:Tính tổng bình phương các chữ số của $n$.Lặp lại bước đó với kết quả vừa tính.Nếu cuối cùng kết quả là 1, thì $n$ là số hạnh phúc.Nếu rơi vào vòng lặp vô hạn không bao giờ ra 1 (ví dụ như $4 \rightarrow 16 \rightarrow 37 \rightarrow \dots$), thì $n$ không phải là số hạnh phúc.
#### Hãy nhập thử đầu vào là "a" rồi đọc kỹ Traceback để gỡ lỗi nhập liệu. (Dùng try...except bắt lỗi ValueError thêm set).

```python
try:
    n = int(input())
    s = set()  
    is_happy = True
    while n != 1:
        if n in s:
            is_happy = False
            break
        s.add(n)
        tong_binh_phuong = 0
        for chu_so in str(n):
            tong_binh_phuong += int(chu_so) ** 2
        n = tong_binh_phuong
    if is_happy:
        print("La so hanh phuc")
    else:
        print("Khong phai so hanh phuc")

except ValueError:
    print("Lỗi dữ liệu đầu vào không phải là số nguyên!")
```

#### W10A4: Sử dụng try...except kết hợp cùng else... finally tính tổng 2 số, nếu nhập sai kiểu thì thông báo. Luôn luôn in "Kết thúc chương trình." sau khi chạy.

```python
try:
    x = float(input())
    y = float(input())
    tong = x + y
    print(f"Tổng: {tong}")
except ValueError:
    print("Thông báo: Nhập sai kiểu dữ liệu!")
else:
    print("Phép tính thực hiện thành công.")
finally:
    print("Kết thúc chương trình.")
```

#### W10A5: Chuyển đổi list sang int: Viết chương trình chuyển tất cả phần tử của một list cho trước về int. Nếu phần tử nào trong list không thể chuyển được thì in cảnh báo và bỏ qua phần tử đó.

```python
danh_sach_goc = ["10", "20", "abc", "30.5", "40"]
danh_sach_so_nguyen = []
for i in danh_sach_goc:
    try:
        so = int(i)
        danh_sach_so_nguyen.append(so)
    except ValueError:
        print(f"Không thể chuyển '{i}' thành số nguyên. Bỏ qua!")
print("List sau khi chuyển:", danh_sach_so_nguyen)
```

#### W10A6: Tính căn bậc hai
#### Đề bài: Viết chương trình nhập một số $x$ và tính căn bậc hai của nó. Bắt lỗi ngoại lệ khi $x$ là số âm. Kết quả in ra chính xác tới 2 chữ số thập phân nếu $x$ hợp lệ. (Dùng try...except bắt lỗi ValueError in ra "So am").

```python
import math
try:
    x = float(input())
    if x < 0:
        raise ValueError("So am") # Chủ động ném lỗi nếu là số âm
    ket_qua = math.sqrt(x)
    print(f"{ket_qua:.2f}")
except ValueError as e:
    if str(e) == "So am":
        print("So am")
    else:
        print("Dữ liệu nhập vào không hợp lệ!")
```

#### W10A10: Kiểm tra mảng hợp lệ
#### Đề bài: Viết chương trình nhập vào phần tử $N$ ($N > 0$) và một mảng gồm $N$ số nguyên dương. Một mảng là hợp lệ nếu nó gồm các số khác nhau. In ra tổng các phần tử nếu mảng đó là hợp lệ. Nếu mảng không hợp lệ, in ra "Mang khong hop le".

```python
try:
    N = int(input())
    if N <= 0:
        raise ValueError
    mang = []
    for i in range(N):
        mang.append(int(input()))
    if len(mang) == len(set(mang)):
        print(sum(mang))
    else:
        print("Mang khong hop le")
except ValueError:
    print("Mang khong hop le")
```

#### W10A13 IndexError - PopElement: Nhập vào một danh sách số nguyên và một chỉ số $k$. Hãy in ra phần tử bị xóa khi thực hiện pop(k). Nếu $k$ không hợp lệ, in ra "Chi so khong hop le".

```python
try:
    danh_sach = list(map(int, input().split()))
    k = int(input())
    phan_tu_bi_xoa = danh_sach.pop(k)
    print(phan_tu_bi_xoa)
except IndexError:
    print("Chi so khong hop le")
except ValueError:
    print("Dữ liệu nhập vào sai định dạng!")
```

#### W10A14 TypeError - SumNumbers: Nhập vào một danh sách gồm nhiều giá trị. Hãy cộng tất cả phần tử sau khi chuyển được sang int. Nếu phần tử nào không chuyển được thì bỏ qua. In ra tổng cuối cùng.

```python
danh_sach_vao = input().split()
tong = 0
for i in danh_sach_vao:
    try:
        so = int(i)
        tong += so
    except (ValueError, TypeError):
        continue
print(tong)
```

#### W10A15 Raise_Exception - PositiveNumber: Nhập vào một số nguyên $n$. Nếu $n \le 0$ thì chủ động phát sinh ngoại lệ và in ra "So phai duong". Nếu hợp lệ thì in ra bình phương của $n$.

```python
try:
    n = int(input())
    if n <= 0:
        raise ValueError("So phai duong")  
    print(n ** 2)
except ValueError as e:
    if str(e) == "So phai duong":
        print("So phai duong")
    else:
        print("Vui lòng nhập một số nguyên!")
```
