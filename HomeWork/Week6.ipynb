#### W6A4

```python
k_v = input().split()
res = {}
for i in k_v:
    k, v = i.split(':')
    if k not in res:
        res[k] = []
    res[k].append(v)
print(res)
```

#### W6A5

```python
nums = list(map(int, input().split()))
res = {'positives':0, 'negatives':0, 'zeros':0}
for i in nums:
    if i > 0:
        res['positives'] += 1
    elif i < 0:
        res['negatives'] += 1
    else:
        res['zeros'] += 1
print(res)
```

#### W6A9

```python

def get_dict():
    return {k: int(v) for item in input().split() for k, v in [item.split(':')]}

d1 = get_dict()
d2 = get_dict()
for k, v in d2.items():
    d1[k] = d1.get(k, 0) + v

# In dict đã sắp xếp theo key
print(dict(sorted(d1.items())))
```

#### W6A10

```python
nums = map(int, input().split())
k = int(input())
seen = set()
ans = set()
for i in nums:
    if k - i in seen:
        ans.add((k - i, i))
    seen.add(i)
ans = sorted(list(ans))
print(ans)
```

#### W6A12

```python
nums = list(map(int, input().split()))
cnt = {}
for i in nums: cnt[i] = cnt.get(i, 0) + 1
max_cnt = max(cnt.values())
ans = [i for i, cnt in cnt.items() if cnt == max_cnt]
print(min(ans))
```
