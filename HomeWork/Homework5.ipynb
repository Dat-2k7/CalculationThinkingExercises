### W5A16: Viết hàm kiểm tra một số có phải số Armstrong hay không (tổng các chữ số lũy thừa với số chữ số bằng chính nó). Trả về kiểu bool, tham số là số nguyên x.

#### Ví dụ 153 là số Armstrong tức $1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153$.

```python
def is_armstrong(x):
    s = str(x) # Chuyển số sang chuỗi
    n = len(s) # Đếm số kí tự trong chuỗi
    return sum(int(i)**n for i in s) == x
assert is_armstrong(153) == True
assert is_armstrong(9474) == True
assert is_armstrong(123) == False
assert is_armstrong(370) == True
assert is_armstrong(5) == True
print("All test acepted")
```

### W5A17: Viết hàm kiểm tra một số có phải số chính phương hay không. Trả về bool, tham số là số nguyên x.

#### ví dụ: $9 = 3^2, 16 = 4^2$

```python
import math
def so_cp(x):
    if x < 0: return False
    a = int(math.isqrt(x))
    return a * a == x
assert so_cp(9) == True
assert so_cp(16) == True
assert so_cp(8) == False
assert so_cp(25) == True
assert so_cp(0) == True
print("All test acepted")
```

### W5A18: Viết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.

#### ví dụ: 121, 55, 12321

```python
def is_palindrome(n):
    if n < 0: return False
    return str(n) == str(n)[::-1]
assert is_palindrome(121) == True
assert is_palindrome(55) == True
assert is_palindrome(123) == False
assert is_palindrome(12321) == True
assert is_palindrome(-121) == False
print("All test acepted")
```

#### Ví dụ: $19 \rightarrow 1+9 = 10 \rightarrow 1+0 = 1$.

### W5A20: Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại trả về 'neither'.

```python
def detect_number_type(s):
    temp_s = s[1:] if s.startswith('-') else s
    if '.' in temp_s and temp_s.replace('.', '', 1).isdigit():
        return 'float'
    if temp_s.isdigit():
        return 'int'
    return 'neither'
assert detect_number_type("123") == 'int'
assert detect_number_type("-123") == 'int'
assert detect_number_type("12.3") == 'float'
assert detect_number_type("-12.3") == 'float'
assert detect_number_type("abc") == 'neither'
print("All test acepted")
```

### W5A22: Hàm sum_halves(s) nhận chuỗi 6 chữ số, trả về tổng của 3 chữ số đầu và 3 chữ số cuối.

```python
def sum_halves(s):
    s = str(s)
    ba_so_dau = sum(int(d) for d in s[:3])
    ba_so_sau = sum(int(d) for d in s[3:])
    return ba_so_dau, ba_so_sau
assert sum_halves("123456") == (6, 15)
assert sum_halves("111111") == (3, 3)
assert sum_halves("000999") == (0, 27)
assert sum_halves(123123) == (6, 6)
assert sum_halves("505010") == (10, 1)
print("All test acepted")
```

### W5A23: Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình $ax^2 + bx + c = 0$.

```python
def quadratic_roots_count(a, b, c):
    if a == 0:
        if b != 0: return 1
        return float('inf') if c == 0 else 0
    delta = b**2 - 4*a*c
    
    if delta > 0:
        return 2
    elif delta == 0:
        return 1
    else:
        return 0
assert quadratic_roots_count(1, -3, 2) == 2   # Delta = 1
assert quadratic_roots_count(1, 2, 1) == 1    # Delta = 0
assert quadratic_roots_count(1, 1, 1) == 0    # Delta = -3
assert quadratic_roots_count(0, 2, 4) == 1    # PT bac 1: 2x + 4 = 0
assert quadratic_roots_count(1, 6, 5) == 2    # Giong vi du trong anh
print("All test acepted")
```

### W5A24: Viết hàm are_collinear(x1, y1, x2, y2, x3, y3) kiểm tra ba điểm có thẳng hàng không.

#### Độ dốc của một đường thẳng đi qua hai điểm $A(x_1, y_1)$ và $B(x_2, y_2)$: $$m_{AB} = \frac{y_2 - y_1}{x_2 - x_1}$$

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (y2 - y1) * (x3 - x2) == (y3 - y2) * (x2 - x1)
assert are_collinear(1, 2, 2, 4, 3, 6) == True
assert are_collinear(0, 0, 1, 1, 2, 2) == True
assert are_collinear(0, 0, 1, 1, 2, 3) == False
assert are_collinear(1, 1, 5, 1, 10, 1) == True  
assert are_collinear(2, 2, 2, 5, 2, 10) == True  
print("All test acepted")
```

### W5A27: Đếm số lượng chữ số khác 0 trong một danh sách các số nguyên nhập vào.

```python
def dem_chu_so_khac_khong(danh_sach_so):
    dem = 0
    for so in danh_sach_so:
        for chu_so in str(abs(so)):
            if chu_so != '0':
                dem += 1
    return dem
assert dem_chu_so_khac_khong([102, 300, 45]) == 5
assert dem_chu_so_khac_khong([0, 0, 0]) == 0       
assert dem_chu_so_khac_khong([123, -405]) == 5    
assert dem_chu_so_khac_khong([7]) == 1
assert dem_chu_so_khac_khong([]) == 0            
print("All test acepted")
```

### W5A28: Tính góc $\angle ABC$ (độ) từ tọa độ 3 điểm A, B, C.

```python
import math

def calculate_angle(x1, y1, x2, y2, x3, y3):
    ba = (x1 - x2, y1 - y2)
    bc = (x3 - x2, y3 - y2)
    dot_product = ba[0]*bc[0] + ba[1]*bc[1]
    mag_ba = math.sqrt(ba[0]**2 + ba[1]**2)
    mag_bc = math.sqrt(bc[0]**2 + bc[1]**2)
    angle = math.acos(dot_product / (mag_ba * mag_bc))
    return math.degrees(angle)
```

### W5A33: Tính tiền taxi dựa trên loại xe, giờ bắt đầu, số km, phút chờ...

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    # Giả định một biểu giá cơ bản
    base_fare = 15000 if car_type == '4-seat' else 20000
    km_rate = 12000
    total = base_fare + (km * km_rate) + (wait_minutes * 500)
    if start_hour >= 22 or start_hour <= 5: # Phụ phí đêm
        total *= 1.1
    return total
```

### W5A34: Tính tổng các số $x$ từ 1 đến $n$ mà các chữ số của $x$ không lặp lại.

```python
def sum_1_to_n(n):
    if n <= 0: return 0
    total = 0
    for x in range(1, n + 1):
        s = str(x)
        if len(set(s)) == len(s): # Kiểm tra tính duy nhất của chữ số
            total += x
    return total
```

```python
def is_prime_number(n):
    if n < 2: return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True
assert is_prime_number(3) == True
assert is_prime_number(2) == True
assert is_prime_number(5) == True
assert is_prime_number(-10) == False
assert is_prime_number(9) == False
print("All test accepted")
```

```python
def is_perfect_number(n):
    if n < 2: return False
    sum = 0
    for i in range(1, n):
        if n % i == 0:
            sum += i
    return sum  == n
assert is_perfect_number(6) == True
assert is_perfect_number(9) == False
assert is_perfect_number(28) == True
assert is_perfect_number(496) == True
assert is_perfect_number(1) == False
print("Pass all test")
```

```python
def ten_To_Bin(n: int) -> str:
    ans = ""
    while n >0:
        du = n % 2
        ans = chr(du + ord("0")) +ans 
        n //=2
    return ans
print(ten_To_Bin(10))
```
