#### W8A1: Cylinder (Lớp hình trụ)
#### Đề bài: Xây dựng lớp Cylinder (hình trụ) gồm: các thuộc tính bán kính và chiều cao (kiểu số thực) với hai phương thức: tính diện tích bề mặt (diện tích toàn phần) và thể tích của hình trụ. (Sử dụng lại những hàm vừa viết, quy ước $\pi = 3.14$).Đầu vào: Một dòng duy nhất chứa bán kính và chiều cao của hình trụ, cách nhau bởi một dấu cách.Đầu ra: In ra màn hình diện tích bề mặt và thể tích của hình trụ (Kết quả làm tròn đến 2 chữ số sau dấu phẩy).

```python
class Cylinder:
    def __init__(self, r, h):
        self.r = r
        self.h = h
    def dt_be_mat(self):
        pi = 3.14
        return 2 * pi * self.r * self.h + 2 * pi * (self.r ** 2)
    def the_tich(self):
        pi = 3.14
        return pi * (self.r ** 2) * self.h
try:
    r, h = map(float, input().split())
    cylinder = Cylinder(r, h)
    print(f"{cylinder.dt_be_mat():.2f} {cylinder.the_tich():.2f}")
except:
    pass
```

#### W8A2: Date (Lớp ngày tháng)
#### Đề bài: Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương thức: kiểm tra xem một ngày tháng nào đó có hợp lệ hay không và phương thức tính ngày kế tiếp của ngày tháng năm nào đó nếu nó là hợp lệ.Nếu ngày tháng năm không tồn tại, in ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó theo định dạng dd/mm/yyyy.Đầu vào: Chứa dữ liệu ngày, tháng, năm dưới dạng dd/mm/yyyy.Đầu ra: In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng dd/mm/yyyy. Nếu ngày tháng không tồn tại, in ra INVALID.

```python
class Date:
    def __init__(self, date_str):
        self.date_str = date_str
        self.day = 0
        self.month = 0
        self.year = 0
        self.is_valid = self.parse_and_validate()
    def is_leap_year(self, y):
        return (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0)
    def get_max_days(self, m, y):
        if m in [1, 3, 5, 7, 8, 10, 12]:
            return 31
        if m in [4, 6, 9, 11]:
            return 30
        if m == 2:
            return 29 if self.is_leap_year(y) else 28
        return 0
    def parse_and_validate(self):
        try:
            parts = self.date_str.split('/')
            if len(parts) != 3:
                return False
            self.day = int(parts[0])
            self.month = int(parts[1])
            self.year = int(parts[2])
            
            if self.year < 1 or self.month < 1 or self.month > 12:
                return False
            max_days = self.get_max_days(self.month, self.year)
            if self.day < 1 or self.day > max_days:
                return False
            return True
        except:
            return False
    def next_date(self):
        if not self.is_valid:
            return "INVALID"
        d, m, y = self.day, self.month, self.year
        max_days = self.get_max_days(m, y)
        d += 1
        if d > max_days:
            d = 1
            m += 1
            if m > 12:
                m = 1
                y += 1   
        return f"{d:02d}/{m:02d}/{y:04d}"
try:
    inp = input().strip()
    date_obj = Date(inp)
    print(date_obj.next_date())
except:
    pass
```

#### W8A3: Point (Lớp điểm tọa độ): Xây dựng lớp điểm với hai thuộc tính $x, y$ cùng các phương thức kiểm tra xem có phải gốc tọa độ không, có nằm trên trục tung không, có nằm trên trục hoành không, tính khoảng cách từ điểm đó tới một điểm khác.Đầu vào: Nhập vào tọa độ một điểm $M$ bất kỳ (gồm 2 số thực $x, y$).Đầu ra: Thông báo $M$ có phải gốc tọa độ không hoặc có nằm trên trục tung hoặc trục hoành không, khoảng cách từ $M$ tới gốc tọa độ là bao nhiêu (lấy 2 chữ số phần thập phân).

```python
import math
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def is_origin(self):
        return self.x == 0 and self.y == 0
    def is_on_y_axis(self):
        return self.x == 0 and self.y != 0
    def is_on_x_axis(self):
        return self.y == 0 and self.x != 0
    def distance_to(self, other):
        return math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)
try:
    mx, my = map(float, input().split())
    M = Point(mx, my)
    O = Point(0, 0)
    if M.is_origin():
        print("M la goc toa do")
    elif M.is_on_y_axis():
        print("M nam tren truc tung")
    elif M.is_on_x_axis():
        print("M nam tren truc hoanh")
    else:
        print("M khong nam tren truc nao dac biet")
    print(f"{M.distance_to(O):.2f}")
except:
    pass
```

#### W8A5 - Lớp ShoppingCart (Giỏ hàng): Xây dựng lớp ShoppingCart để quản lý các sản phẩm của khách hàng. Dùng một danh sách (list) để lưu các sản phẩm, trong đó mỗi sản phẩm là một từ điển (dictionary) gồm các thuộc tính name (tên) và price (giá).
#### Lớp gồm các phương thức sau:
#### - Khởi tạo với danh sách sản phẩm rỗng.
#### - Thêm vào giỏ hàng một sản phẩm mới.
#### - Xóa khỏi giỏ hàng một sản phẩm nào đó (tìm theo tên để xóa).
#### - Kiểm tra xem giỏ hàng có rỗng không.
#### - Tính tổng tiền của giỏ hàng.
#### - Hiển thị toàn bộ sản phẩm trong giỏ hàng.
#### - Xóa toàn bộ giỏ hàng.
#### Yêu cầu tương tác: Tạo menu cho người dùng nhập lựa chọn truy vấn từ 2 tới 7. Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có muốn thoát khỏi chương trình không?".

```python
class ShoppingCart:
    def __init__(self):
        self.products = []
    def add_product(self, name, price):
        self.products.append({
            "name": name,
            "price": price
        })
    def remove_product(self, name):
        for product in self.products:
            if product["name"] == name:
                self.products.remove(product)
                print("Đã xóa sản phẩm.")
                return
        print("Không tìm thấy sản phẩm.")
    def is_empty(self):
        return len(self.products) == 0
    def total_price(self):
        total = 0
        for product in self.products:
            total += product["price"]
        return total
    def show_products(self):
        if self.is_empty():
            print("Giỏ hàng trống.")
        else:
            print("\nDanh sách sản phẩm:")
            for i, product in enumerate(self.products, 1):
                print(f"{i}. {product['name']} - {product['price']}")
    def clear_cart(self):
        self.products.clear()
        print("Đã xóa toàn bộ giỏ hàng.")

cart = ShoppingCart()
while True:
    print("\n===== SHOPPING CART =====")
    print("2. Thêm sản phẩm")
    print("3. Xóa sản phẩm")
    print("4. Kiểm tra giỏ hàng rỗng")
    print("5. Tính tổng tiền")
    print("6. Hiển thị sản phẩm")
    print("7. Xóa toàn bộ giỏ hàng")
    choice = input("Chọn chức năng (2-7): ")
    if choice == "2":
        name = input("Tên sản phẩm: ")
        price = float(input("Giá sản phẩm: "))
        cart.add_product(name, price)
    elif choice == "3":
        name = input("Nhập tên sản phẩm cần xóa: ")
        cart.remove_product(name)
    elif choice == "4":
        if cart.is_empty():
            print("Giỏ hàng đang rỗng.")
        else:
            print("Giỏ hàng không rỗng.")
    elif choice == "5":
        print("Tổng tiền:", cart.total_price())
    elif choice == "6":
        cart.show_products()
    elif choice == "7":
        cart.clear_cart()
    else:
        print("Lựa chọn không hợp lệ.")
    answer = input("\nCó muốn thoát khỏi chương trình không? (yes/no): ")
    if answer.lower() == "yes":
        break
```

 #### W8A6 - Lớp Rectangle (Hình chữ nhật): Xây dựng lớp Rectangle(w, h) với w (chiều rộng) và h (chiều cao) kèm các phương thức:area(): Trả về diện tích.perimeter(): Trả về chu vi.scale(k): Phóng to hình chữ nhật theo hệ số $k$ (nhân cả $w$ và $h$ với $k$), không trả về gì.

```python
class Rectangle:
    def __init__(self, w, h):
        self.w = float(w)
        self.h = float(h)
    def area(self):
        # Diện tích = w * h
        return self.w * self.h
    def perimeter(self):
        # Chu vi = (w + h) * 2
        return (self.w + self.h) * 2
    def scale(self, k):
        # Phóng to kích thước lên k lần
        self.w *= k
        self.h *= k
try:
    w, h, k = map(float, input().split())
    rect = Rectangle(w, h)
    rect.scale(k)
    print(f"{rect.area():.2f}")
    print(f"{rect.perimeter():.2f}")
except Exception as e:
    pass
```

#### W8A7: Lớp Phân Số (Fraction)Đề bài: Thiết kế lớp Fraction để thực hiện các phép toán với phân số. Lớp gồm các thuộc tính numerator (tử số) và denominator (mẫu số). Viết các phương thức để cộng, trừ, nhân, chia hai phân số và phương thức tối giản phân số.Đầu vào: Một dòng gồm 4 số nguyên $a, b, c, d$ ($b, d \neq 0$) lần lượt là tử và mẫu của phân số thứ nhất, tử và mẫu của phân số thứ hai.Đầu ra: In ra kết quả phép cộng, trừ, nhân, chia sau khi đã tối giản dưới dạng tu/mau. Nếu mẫu số bằng 1 thì chỉ in tử số. Nếu kết quả bằng 0 thì in 0. Các kết quả cách nhau bởi một dấu cách.

```python
import math
import sys
class Fraction:
    def __init__(self, num, den):
        if den == 0:
            raise ValueError("Mẫu số phải khác 0")
        self.num = num
        self.den = den
        self.simplify()
    def simplify(self):
        # Tìm ước chung lớn nhất 
        gcd = math.gcd(abs(self.num), abs(self.den))
        self.num //= gcd
        self.den //= gcd
        if self.den < 0:
            self.num = -self.num
            self.den = -self.den
    def add(self, other):
        return Fraction(self.num * other.den + other.num * self.den, self.den * other.den)
    def sub(self, other):
        return Fraction(self.num * other.den - other.num * self.den, self.den * other.den)
    def mul(self, other):
        return Fraction(self.num * other.num, self.den * other.den)
    def div(self, other):
        return Fraction(self.num * other.den, self.den * other.num)
    def __str__(self):
        if self.num == 0:
            return "0"
        if self.den == 1:
            return str(self.num)
        return f"{self.num}/{self.den}"
try:
    line = input().strip()
    parts = line.split()
    if len(parts) == 5:
        a = int(parts[0])
        b = int(parts[1])
        op = parts[2]      
        c = int(parts[3])
        d = int(parts[4])
        f1 = Fraction(a, b)
        f2 = Fraction(c, d)
        if op == '+':
            print(f1.add(f2))
        elif op == '-':
            print(f1.sub(f2))
        elif op == '*':
            print(f1.mul(f2))
        elif op == '/':
            print(f1.div(f2))
except:
    pass
```

#### W8A8 - Lớp BankAccount (Tài khoản ngân hàng): Xây dựng lớp BankAccount(owner, balance) dùng để quản lý tài khoản ngân hàng của một khách hàng với hai thuộc tính: owner (tên chủ tài khoản) và balance (số dư tài khoản).

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = float(balance)
    def deposit(self, x):
        self.balance += x
    def withdraw(self, x):
        if x <= self.balance:
            self.balance -= x
    def __str__(self):
        return f"{self.owner}: {self.balance:.2f}"
owner, balance = input().split()
acc = BankAccount(owner, float(balance))
n = int(input())
for _ in range(n):
    cmd, x = input().split()
    x = float(x)
    if cmd == "DEPOSIT":
        acc.deposit(x)
    elif cmd == "WITHDRAW":
        acc.withdraw(x)
print(f"{acc.balance:.2f}")
```

#### W8A9: Quản lý Học sinh (Student): Xây dựng lớp Student gồm thuộc tính name và danh sách điểm scores. Phương thức add_score(subject, score): Thêm điểm của một môn học. Phương thức avg(): Trả về điểm trung bình của các môn học. Phương thức rank(): Xếp loại học lực dựa trên điểm trung bình (avg): Excellent nếu avg >= 8, Good nếu 6.5 <= avg < 8, Average nếu 5 <= avg < 6.5, ngược lại Poor.Đầu vào: Dòng 1 là tên học sinh. Dòng 2 là số nguyên n (số lượng môn học). n dòng tiếp theo, mỗi dòng gồm: tên_môn_học điểm_số (điểm số là số thực từ 0 đến 10). Đầu ra: In ra trên một dòng gồm: Tên_học_sinh Điểm_trung_bình Xếp_loại (Điểm trung bình làm tròn 2 chữ số thập phân).

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []
    def add_score(self, subject, score):
        self.scores.append(score)
    def avg(self):
        if len(self.scores) == 0:
            return 0
        return sum(self.scores) / len(self.scores)
    def rank(self):
        a = self.avg()
        if a > 8:
            return "Excellent"
        elif a >= 6.5:
            return "Good"
        elif a >= 5:
            return "Average"
        else:
            return "Poor"
name = input()
st = Student(name)
n = int(input())
for _ in range(n):
    subject, score = input().split()
    st.add_score(subject, float(score))
print(f"{st.name} {st.avg():.2f} {st.rank()}")
```

#### W8A10: Quản lý Thư viện (Library): Xây dựng lớp Book chứa thuộc tính title, author, year. Xây dựng lớp Library gồm danh sách sách và các phương thức: add(book) để thêm một cuốn sách vào thư viện, count_by_author(name) để đếm số lượng sách của tác giả name, find_by_year(y) để đếm số lượng sách xuất bản vào năm y. Đầu vào: Dòng 1 là số nguyên n (số lượng câu lệnh thao tác). n dòng tiếp theo nhận một trong ba loại lệnh: ADD title;author;year (thông tin cách nhau bởi dấu chấm phẩy), COUNT author, hoặc COUNTYEAR year. Đầu ra: Với mỗi lệnh COUNT hoặc COUNTYEAR, in ra một số nguyên duy nhất là số lượng sách tương ứng trên một dòng.

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = int(year)
class Library:
    def __init__(self):
        self.books = []
    def add(self, book):
        self.books.append(book)
    def count_by_author(self, name):
        cnt = 0
        for b in self.books:
            if b.author == name:
                cnt += 1
        return cnt
    def find_by_year(self, y):
        cnt = 0
        for b in self.books:
            if b.year == y:
                cnt += 1
        return cnt
lib = Library()
n = int(input())
for _ in range(n):
    line = input()
    if line.startswith("ADD "):
        data = line[4:]
        title, author, year = data.split(";")
        lib.add(Book(title, author, int(year)))
    elif line.startswith("COUNTYEAR "):
        year = int(line.split()[1])
        print(lib.find_by_year(year))
    elif line.startswith("COUNT "):
        author = line.split(maxsplit=1)[1]
        print(lib.count_by_author(author))
```

#### W8A11: Tính Góc Giữa Hai Mặt Phẳng 3D: Cho bốn điểm A, B, C và D trong hệ tọa độ Đề-các 3 chiều. Hãy tính và in ra góc giữa hai mặt phẳng tạo bởi các điểm: mặt phẳng (A, B, C) và mặt phẳng (B, C, D). Góc cần in ra được ký hiệu là PHI, và phải dùng đơn vị độ (degrees), không dùng radian.Đầu vào: Gồm 4 dòng, mỗi dòng chứa 3 số thực cách nhau bằng dấu cách biểu diễn tọa độ X, tọa độ Y, tọa độ Z của một điểm. Đầu ra: In ra góc giữa hai mặt phẳng, chính xác tới hai chữ số thập phân.

```python
from math import sqrt, acos, degrees
class Point3D:
    def __init__(self, x, y, z):
        self.x = float(x)
        self.y = float(y)
        self.z = float(z)
def vector(a, b):
    return (
        b.x - a.x,
        b.y - a.y,
        b.z - a.z
    )
def cross(u, v):
    return (
        u[1]*v[2] - u[2]*v[1],
        u[2]*v[0] - u[0]*v[2],
        u[0]*v[1] - u[1]*v[0]
    )
def dot(u, v):
    return u[0]*v[0] + u[1]*v[1] + u[2]*v[2]
def norm(v):
    return sqrt(dot(v, v))
A = Point3D(*map(float, input().split()))
B = Point3D(*map(float, input().split()))
C = Point3D(*map(float, input().split()))
D = Point3D(*map(float, input().split()))
AB = vector(A, B)
AC = vector(A, C)
BC = vector(B, C)
BD = vector(B, D)
n1 = cross(AB, AC)
n2 = cross(BC, BD)
cos_phi = abs(dot(n1, n2)) / (norm(n1) * norm(n2))
phi = degrees(acos(cos_phi))
print(f"{phi:.2f}")
```

#### W8A12: Phép Nhân Số Phức (ComplexNumber) : Khai báo kiểu dữ liệu biểu diễn số phức ComplexNumber bao gồm 2 biến realPart và imaginaryPart kiểu double biểu diễn phần thực và phần ảo của một số phức. Viết hàm khởi tạo ComplexNumber(double initRealPart, double initImaginaryPart) và hàm ComplexNumber multiply(complexNumber other) trả về tích của số phức hiện tại và một số phức khác. Đầu vào: Đầu vào từ bàn phím gồm hai dòng, mỗi dòng gồm hai số thực cách nhau bởi một dấu cách biểu diễn phần thực và phần ảo của một số phức. Các số trong đầu vào có giá trị tuyệt đối không vượt quá 100. Đầu ra: In ra màn hình hai số biểu diễn phần thực và phần ảo của tích hai số phức.

```python
class ComplexNumber:
    def __init__(self, realPart, imaginaryPart):
        self.real = realPart
        self.imag = imaginaryPart
    def multiply(self, other):
        real = self.real * other.real - self.imag * other.imag
        imag = self.real * other.imag + self.imag * other.real
        return ComplexNumber(real, imag)
a, b = map(float, input().split())
c, d = map(float, input().split())
z1 = ComplexNumber(a, b)
z2 = ComplexNumber(c, d)
result = z1.multiply(z2)
print(f"{result.real:g} {result.imag:g}")
```

#### W8A13: Tính Diện Tích Tam Giác (Triangle): Khai báo cấu trúc dữ liệu tam giác Triangle chứa 3 biến edge1, edge2, edge3 kiểu int là độ dài 3 cạnh của tam giác đó. Viết hàm khởi tạo Triangle(int initEdge1, int initEdge2, int initEdge3) để khởi tạo độ dài các cạnh và hàm double getArea() trả về diện tích của tam giác đó (nếu tam giác không hợp lệ, trả về -1). Đầu vào: Một dòng duy nhất từ bàn phím chứa 3 số nguyên là độ dài 3 cạnh của tam giác. Đầu ra: In ra màn hình diện tích tam giác (làm tròn đến 2 chữ số thập phân sau dấu phẩy). In ra invalid nếu 3 số nhập vào không phải là 3 cạnh của một tam giác.

```python
from math import sqrt
class Triangle:
    def __init__(self, edge1, edge2, edge3):
        self.a = edge1
        self.b = edge2
        self.c = edge3
    def getArea(self):
        if (self.a + self.b <= self.c or
            self.a + self.c <= self.b or
            self.b + self.c <= self.a):
            return -1
        p = (self.a + self.b + self.c) / 2
        return sqrt(
            p *
            (p - self.a) *
            (p - self.b) *
            (p - self.c)
        )
a, b, c = map(int, input().split())
tri = Triangle(a, b, c)
area = tri.getArea()
if area == -1:
    print("invalid")
else:
    if area.is_integer():
        print(int(area))
    else:
        print(f"{area:.2f}")
```

#### W8A14: Kế thừa Tiếng Kêu Động Vật (Animal): Xây dựng lớp Animal với phương thức speak(). Xây dựng hai lớp kế thừa Dog và Cat, mỗi lớp ghi đè speak() để trả về âm thanh tương ứng: Dog -> Woof, Cat -> Meow. Đầu vào: Một dòng chứa tên lớp: Dog hoặc Cat. Đầu ra: In ra kết quả của phương thức speak() tương ứng với lớp động vật đó.

```python
class Animal:
    def speak(self):
        pass  
class Dog(Animal):
    def speak(self):
        return "Woof"
class Cat(Animal):
    def speak(self):
        return "Meow"
animal_type = input().strip()
if animal_type == "Dog":
    obj = Dog()
elif animal_type == "Cat":
    obj = Cat()
else:
    obj = None
if obj:
    print(obj.speak())
```

#### W8A15: Quản lý Thu nhập Nhân viên (Employee & Manager): Xây dựng lớp Employee(name, salary) và lớp Manager(name, salary, bonus) kế thừa từ Employee. Cả hai lớp đều có phương thức income(): Employee trả về salary, Manager trả về salary + bonus. Đầu vào: Dòng 1 là loại đối tượng (Employee hoặc Manager). Dòng 2 chứa name, salary. Nếu dòng 1 là Manager thì dòng này sẽ có thêm chỉ số bonus. Đầu ra: In ra một số thực duy nhất là tổng thu nhập của đối tượng nhận được từ phương thức income().

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = float(salary)
    def income(self):
        return self.salary
class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = float(bonus)
    def income(self):
        return self.salary + self.bonus
role = input().strip() 
data = input().split()  
if role == "Employee":
    name = data[0]
    salary = data[1]
    emp = Employee(name, salary)
    print(f"{emp.income():.0f}")  
elif role == "Manager":
    name = data[0]
    salary = data[1]
    bonus = data[2]
    mgr = Manager(name, salary, bonus)
    print(f"{mgr.income():.0f}")
```
