### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh
1 3 5 6 7 8 9: Tang
9 8 7 6 5 4 0: Giam
1 2 3 7 1 0 2: Vo danh
1 2 2 4 5 7: Vo danh

```python
def check_sequence(arr):
    if len(arr) < 2:
        return "Tang" # Hoặc tùy quy ước cho dãy 1 phần tử

    is_increasing = True
    is_decreasing = True

    for i in range(len(arr) - 1):
        if arr[i] >= arr[i+1]:
            is_increasing = False
        if arr[i] <= arr[i+1]:
            is_decreasing = False

    if is_increasing:
        return "Tang"
    elif is_decreasing:
        return "Giam"
    else:
        return "Vo danh"
print(f"1 3 5 6 7 8 9: {check_sequence([1, 3, 5, 6, 7, 8, 9])}")
print(f"9 8 7 6 5 4 0: {check_sequence([9, 8, 7, 6, 5, 4, 0])}")
print(f"1 2 3 7 1 0 2: {check_sequence([1, 2, 3, 7, 1, 0, 2])}")
print(f"1 2 2 4 5 7: {check_sequence([1, 2, 2, 4, 5, 7])}") 
```

### Bài 8: Kiểm tra một list số nguyên có phải là ziczac khong
1 3 2 5 2 8 7 9 2 100: ziczac
34 2 56 7 90 20 30 7 120: ziczac

```python
def is_zigzag(arr):
    if len(arr) < 3:
        return "ziczac" # Quy ước dãy ngắn là ziczac

    for i in range(1, len(arr) - 1):
        # Kiểm tra nếu 3 số liên tiếp không tạo thành hình chữ V hoặc V ngược
        # Tức là nếu nó cứ tăng mãi hoặc giảm mãi tại một điểm nào đó
        check_1 = (arr[i-1] < arr[i] and arr[i] > arr[i+1]) # Cực đại
        check_2 = (arr[i-1] > arr[i] and arr[i] < arr[i+1]) # Cực tiểu
        
        if not (check_1 or check_2):
            return "khong phai ziczac"
            
    return "ziczac"

# Test cases
test1 = [1, 3, 2, 5, 2, 8, 7, 9, 2, 100]
test2 = [34, 2, 56, 7, 90, 20, 30, 7, 120]

print(f"{test1}: {is_zigzag(test1)}")
print(f"{test2}: {is_zigzag(test2)}")
```

### Bài 9: Tìm và in ra tất cả các vị trí của số nguyên k trong một danh sách số nguyên cho trước. Nếu không có k trong ds thì in ra "-1".
VD: k = 3, [1,3,4,4,3,3,4,3,6,1] => k xuất hiện tại các vị trí (trả về 1 list) [1, 4, 5, 7]
k = 10, [1,3,4,4,3,3,4,3,6,1] => k không xuất hiện trong ds => trả về "-1"

```python
def find_all_indices(lst, k):
    indices = [i for i, x in enumerate(lst) if x == k]
    return indices if indices else -1
lst_test = [1, 3, 4, 4, 3, 3, 4, 3, 6, 1]
print(f"k = 3: {find_all_indices(lst_test, 3)}")   
print(f"k = 10: {find_all_indices(lst_test, 10)}") 
```
