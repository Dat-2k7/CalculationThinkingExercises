#### W6A16: In ma trận căn phải 4 vị trí

```python
n, m = map(int, input("Nhập n và m: ").split())
matrix = []
for i in range(n):
    row = list(map(int, input(f"Nhập hàng {i}: ").split()))
    matrix.append(row)
# In ma trận định dạng 4 vị trí căn phải
for row in matrix:
    for val in row:
        print(f"{val:>4}", end="")
    print()
```

#### W6A17: In đường chéo chính và phụ

```python
n = int(input("Nhập n (ma trận vuông n x n): "))
matrix = [list(map(int, input().split())) for _ in range(n)]
main_diag = [matrix[i][i] for i in range(n)]
side_diag = [matrix[i][n-1-i] for i in range(n)]
print(*(main_diag))
print(*(side_diag))
```

#### W6A18: Tổng các phần tử trên cột k (Lưu ý: cột k tính từ 0)

```python
n, m, k = map(int, input("Nhập n, m, k: ").split())
matrix = []
for i in range(n):
    row = list(map(int, input(f"Nhập hàng {i}: ").split()))
    matrix.append(row)
# Tính tổng cột k (k tính từ 0)
col_sum = sum(row[k] for row in matrix)
print(f"Tổng cột {k} là: {col_sum}")
```

#### W6A21: Dict ánh xạ chuỗi -> tuple các vị trí

```python
s_list = input("Nhập các chuỗi: ").split()
d = {}
for i, s in enumerate(s_list):
    if s not in d:
        d[s] = []
    d[s].append(i)
# Chuyển list vị trí thành tuple
result = {k: tuple(v) for k, v in d.items()}
print(result)
```

#### W6A24: Đẩy số 0 về đầu

#### W6A26: Tổng cực đại địa phương (a[i-1] < a[i] > a[i+1])


```python
lst = list(map(int, input("Nhập danh sách: ").split()))
total = 0
for i in range(1, len(lst) - 1):
    if lst[i-1] < lst[i] > lst[i+1]:
        total += lst[i]
print(total)
```

#### W6A27: Vị trí các số < x

```python
lst = list(map(int, input("Nhập danh sách: ").split()))
x = int(input("Nhập x: "))
indices = [i for i, val in enumerate(lst) if val < x]
if indices:
    print(indices)
else:
    print(-1)
```

#### W6A28: Kiểm tra tăng/giảm thực sự

```python
lst = list(map(int, input("Nhập danh sách: ").split()))
is_increasing = all(lst[i] < lst[i+1] for i in range(len(lst)-1))
is_decreasing = all(lst[i] > lst[i+1] for i in range(len(lst)-1))
if is_increasing:
    print("Tang")
elif is_decreasing:
    print("Giam")
else:
    print("Vo danh")
```
