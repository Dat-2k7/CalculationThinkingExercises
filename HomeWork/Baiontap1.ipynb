#### Bài 1: Cước phí Taxi Thông Minh: Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) để tính tổng tiền cước (VND, kiểu số nguyên).Giá mở cửa (0 - 0.8 km):
##### - Xe 4 chỗ: 11,000 VND.Xe 7 chỗ: 12,000 VND.Giá theo km sau 0.8 km:Xe 4 chỗ: 0.8 đến 20 km tính 13,000 VND/km; trên 20 km tính 10,500 VND/km.
##### - Xe 7 chỗ: 0.8 đến 20 km tính 14,000 VND/km; trên 20 km tính 12,000 VND/km.
##### - Phí chờ: 5 phút đầu miễn phí. Sau đó, mỗi block 5 phút (làm tròn lên) tính 2,000 VND.Phụ thu ban đêm: 
##### - Nếu start_hour trong khoảng $[22, 23, 0, 1, 2, 3, 4, 5]$, cộng thêm 20% vào tiền tạm tính (mở cửa + đường dài + chờ).Điều kiện đặc biệt:Nếu $km > 30$ và không phải ban đêm: Giảm 10% tổng tiền.
##### - Nếu tổng tiền cuối cùng $< 50,000$ nhưng $km > 5$: Đặt tổng tiền bằng 50,000 VND (mức tối thiểu).

```python
import math
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 4:
        open_price = 11000
        price_under_20 = 13000
        price_over_20 = 10500
    elif car_type == 7:
        open_price = 12000
        price_under_20 = 14000
        price_over_20 = 12000
    else:
        return "Không có xe đó"
    #Tính tiền quãng đường đi được
    tien_qd = 0
    if km <= 0.8:
        tien_qd = open_price
    elif km <= 20:
        tien_qd = open_price + (km - 0.8)*price_under_20
    else: #km>20
        tien_qd = open_price + (20 - 0.8)*price_under_20 + (km-20)*price_over_20
    #Tính tiền chờ
    tien_cho = 0
    if wait_minutes > 5:
        tien_cho = math.ceil((wait_minutes-5)/5)*2000
    #Tong tien tam tinh và phụ thu thêm
    sum = tien_qd + tien_cho
    night_hour = [22, 23, 0, 1,2,3,4,5]
    is_night = start_hour in night_hour
    if is_night:
        sum = sum * 1.2
    #Điều kiện đặc biệt và giảm giá
    #Giảm 10% nếu km > 30 và đi vào ban ngày
    if km > 30 and not is_night:
        total = sum*0.9
    else:
        total = sum
    #Kiểm tra mức giá tối thiểu 50000 VND nếu quãng đường > 5 km
    if total < 50000 and km > 5:
        total = 50000
    return int(total)
print("Hệ thống tính cước Taxi thông minh")
try:
    car_type = int(input())
    start_hour = int(input())
    km = float(input())
    wait_minutes = int(input())
    result = smart_taxi_fare(car_type,start_hour,km,wait_minutes)
    print(f"Tổng số tiền phải trả là: {result:,}")
except ValueError:
    print("Vui lòng nhập lại")
```

```python
def check_sum(x):
    if x < 1:
        return False
    tong = 0
    n = 1
    while tong < x:
        tong += n
        if tong == x:
            return True
        n += 1
    return False
x = int(input())
result = check_sum(x)
print(f"Kết quả {result}")


```

##### Bài 4: Cho các hệ số thực $a, b, c$. Tìm tất cả các nghiệm thực của phương trình bậc hai $ax^2 + bx + c = 0$. Kết quả trả về là một danh sách (list) các nghiệm được sắp xếp theo thứ tự tăng dần.Quy tắc cần cài đặt:Trường hợp $a = 0$:Nếu $b \neq 0$: Phương trình bậc nhất, trả về một nghiệm duy nhất $[-c / b]$.Nếu $b = 0$ và $c = 0$: Vô số nghiệm (không có nghiệm hữu hạn duy nhất), trả về danh sách rỗng [].Nếu $b = 0$ và $c \neq 0$: Vô nghiệm, trả về [].Trường hợp $a \neq 0$: Tính biệt thức $\Delta = b^2 - 4ac$.Nếu $\Delta < 0$: Phương trình vô nghiệm thực, trả về [].Nếu $\Delta = 0$: Phương trình có nghiệm kép, trả về $[-b / (2a)]$.Nếu $\Delta > 0$: Phương trình có 2 nghiệm phân biệt $[r1, r2]$ với $r1 < r2$.

```python
import math
def solve_quadratic(a,b,c):
    if a == 0:
        if b != 0:
            return [round(-c/b,1)]
        else:
            return []
    delta = b**2 - 4*a*c
    if delta < 0:
        return []
    elif delta == 0:
        return [-b/(2*a)]
    else:# delta > 0, PT có 2 no
        can_delta = math.sqrt(delta)
        x1 = (-b-can_delta)/(2*a)
        x2 = (-b+can_delta)/(2*a)
        result = [x1, x2]
        result.sort()
        return result
try:
    a, b, c = map(int,input().split())
    nghiem = solve_quadratic(a,b,c)
    print(f"Nghiệm của phương trình là {nghiem}") #{x:.1f}
except:
    print("Lỗi")



```

##### Bài 30: Viết hàm detect_number_type(s) trả về 'int' nếu s biểu diễn một số nguyên, 'float' nếu biểu diễn một số thực dấu chấm động, ngược lại trả về 'neither'. 


```python
import re
def detect_number_type(s):
    if re.fullmatch(r'[+-]?\d+',s):
        return 'int'
    elif re.fullmatch(r'[+-]?\d+\.\d+',s):
        return 'float'
    else:
        return 'neither'
s = input()
print(detect_number_type(s))

```

##### Bài 31: is_perfect_square(n) Kiểm tra số chính phương

```python
import math
def is_perfect_square(n):
    if n < 0:
        return False
    can_bac_hai = int(math.sqrt(n))
    if can_bac_hai * can_bac_hai == n:
        return True
    else:
        return False
n = float(input())
print(is_perfect_square(n))
```

```python
def is_right_triangle(a,b,c):
    if not (a+b>c and a+c>b and b+c>a):
        return False
    canh = sorted([a,b,c])
    if canh[0]**2+canh[1]**2 == canh[2]**2:
        return True
    else: 
        return False
a,b,c = map(int,input().split())
print(is_right_triangle(a,b,c))
```

##### Bài 41: Ký tự đứng sau $n$ vị trí (Xoay vòng)

##### Bài 43: Tính diện tích tam giác (Công thức Heron)

```python
import math
def dien_tich(a,b,c): 
    if not (a+b>c and a+c>b and b+c>a):
        raise ValueError
    p = (a+b+c)/2
    return math.sqrt(p*(p-a)*(p-b)*(p-c))
print(dien_tich(3,4,5))
```

```python
def so_nguyen_to(n):
    if n < 2:
        return False
    for i in range (2, int(n**0.5)+1):
        if n % i ==0:
            return False
    return True
print(so_nguyen_to(5))
```

```python
def triangle(a,b,c):
    if not(a+b>c and a+c>b and b+c>a):
        return "Không là tam giác"
    canh = sorted([a,b,c])
    is_triangle_vuong = canh[0]**2 + canh[1]**2 == canh[2]**2
    is_triangle_can = (a== b) or (a==c) or (b==c)
    if a==b==c:
        return "Tam giác đều"
    elif is_triangle_vuong and is_triangle_can:
        return "Tam giác vuông cân"
    elif is_triangle_vuong:
        return "Tam giác vuông"
    elif is_triangle_can:
        return "Tam giác cân"
    elif canh[0]**2 + canh[1]**2 < canh[2]**2:
        return "Tam giác tù"
    else:
        return "Tam giác nhọn"
a,b,c = map(int,input().split())
print(f"3 canh lập trình tam giác: {triangle(a,b,c)}")




```
